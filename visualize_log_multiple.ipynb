```python
import pandas as pd
import os
logs =[
    {"name":"en_c" ,"path":"F:/logs/effnet_clam/20250702_224308_p6_b75"},
    {"name":"en_cs" ,"path":"F:/logs/effnet_clamsimple/20250702_225543_p6_b75"},
    # {"name": "rs_c","path":"F:/logs/resnet_clam/20250702_230514_p6_b75"},
    # {"name":"rs_cs" ,"path":"F:/logs/resnet_clam_simple/20250702_232408_p6_b75"},
    #{"name": "rsx_c","path":"F:/logs/resnext_clam/20250702_234114_p6_b75"},
    {"name": "rsx_cs" ,"path":"F:/logs/resnext_clam_simple/20250702_235347_p6_b75"},
    {"name": "en_c_sig" ,"path":"F:/logs/effnet_clam_sigmoid/20250704_000108_p6_b75"},
    {"name": "en_cs_sig" ,"path":"F:/logs/effnet_clamsimple_sigmoid/20250704_001632_p6_b75"},
    {"name": "rs_c_sig" ,"path":"F:/logs/resnet_clam_sigmoid/20250704_091810_p6_b75"},
    {"name": "rs_cs_sig" ,"path":"F:/logs/resnet_clam_simple_sigmoid/20250704_094207_p6_b75"},
    {"name": "rsx_c_sig" ,"path":"F:/logs/resnext_clam_sigmoid/20250704_100020_p6_b75"},
    # {"name": "rsx_cs_sig" ,"path":"F:/logs/resnext_clam_simple_sigmoid/20250704_101620_p6_b75"}
]
dfs = []
type_of_log = "Validation"
log_file_name = f"{type_of_log.lower()}_log.csv"
for log in logs:
# Load the CSV
    dfs.append(pd.read_csv(os.path.join(log['path'],log_file_name)))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
for i in range( len(logs)):
    plt.plot(dfs[i]["epoch"], dfs[i]["qwk"], label=logs[i]["name"])
plt.xlabel("Epoch")
plt.ylabel("Score")
plt.title(f"{type_of_log} Quadratic Weighted Kappa")
plt.legend()
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
for i in range( len(logs)):
    plt.plot(dfs[i]["epoch"], dfs[i]["loss"], label=logs[i]["name"])
plt.xlabel("Epoch")
plt.ylabel("Score")
plt.title(f"{type_of_log} Loss")
plt.legend()
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
for i in range( len(logs)):
    plt.plot(dfs[i]["epoch"], dfs[i]["weighted_f1"], label=logs[i]["name"])
plt.xlabel("Epoch")
plt.ylabel("Score")
plt.title(f"{type_of_log} Weighted F1")
plt.legend()
plt.grid()
plt.show()
```
